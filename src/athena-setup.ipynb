```
import boto3
import pandas as pd
import sagemaker
from pyathena import connect
```

```
session = boto3.session.Session()
region = session.region_name
sagemaker_session = sagemaker.Session()
bucket = sagemaker_session.default_bucket()

sagemaker_client = boto3.Session().client(service_name="sagemaker", region_name=region)
s3_client = boto3.Session().client(service_name="s3", region_name=region)
```

```
S3_DATA_PATH = f"s3://{bucket}/final_project/data"
!aws s3 cp "../data/DDoS-Friday-no-metadata.parquet" $S3_DATA_PATH/
```

```
ATHENA_DB_NAME = "final_project"
S3_STAGING_DIR = f"s3://{bucket}/final_project/staging"
conn = connect(region_name=region, s3_staging_dir=S3_STAGING_DIR)
statement = f"CREATE DATABASE IF NOT EXISTS {ATHENA_DB_NAME}"
pd.read_sql(statement, conn)
```

```
DATA_TABLE_NAME = "ddos_data"
create_table_statement = f"""
    CREATE EXTERNAL TABLE IF NOT EXISTS {ATHENA_DB_NAME}.{DATA_TABLE_NAME} (
      protocol int,
      flow_duration int,
      total_fwd_packets int,
      total_backward_packets int,
      fwd_packets_length_total int,
      bwd_packets_length_total int,
      fwd_packet_length_max int,
      fwd_packet_length_min int,
      fwd_packet_length_mean double,
      fwd_packet_length_std double,
      bwd_packet_length_max int,
      bwd_packet_length_min int,
      bwd_packet_length_mean double,
      bwd_packet_length_std double,
      flow_bytes_per_s double,
      flow_packets_per_s double,
      flow_iat_mean double,
      flow_iat_std double,
      flow_iat_max int,
      flow_iat_min int,
      fwd_iat_total int,
      fwd_iat_mean double,
      fwd_iat_std double,
      fwd_iat_max int,
      fwd_iat_min int,
      bwd_iat_total int,
      bwd_iat_mean double,
      bwd_iat_std double,
      bwd_iat_max int,
      bwd_iat_min int,
      fwd_psh_flags int,
      bwd_psh_flags int,
      fwd_urg_flags int,
      bwd_urg_flags int,
      fwd_header_length int,
      bwd_header_length int,
      fwd_packets_per_s double,
      bwd_packets_per_s double,
      packet_length_min int,
      packet_length_max int,
      packet_length_mean double,
      packet_length_std double,
      packet_length_variance double,
      fin_flag_count int,
      syn_flag_count int,
      rst_flag_count int,
      psh_flag_count int,
      ack_flag_count int,
      urg_flag_count int,
      cwe_flag_count int,
      ece_flag_count int,
      down_up_ratio double,
      avg_packet_size double,
      avg_fwd_segment_size double,
      avg_bwd_segment_size double,
      fwd_avg_bytes_bulk double,
      fwd_avg_packets_bulk double,
      fwd_avg_bulk_rate int,
      bwd_avg_bytes_bulk double,
      bwd_avg_packets_bulk double,
      bwd_avg_bulk_rate int,
      subflow_fwd_packets int,
      subflow_fwd_bytes int,
      subflow_bwd_packets int,
      subflow_bwd_bytes int,
      init_fwd_win_bytes int,
      init_bwd_win_bytes int,
      fwd_act_data_packets int,
      fwd_seg_size_min int,
      active_mean double,
      active_std double,
      active_max int,
      active_min int,
      idle_mean double,
      idle_std double,
      idle_max int,
      idle_min int,
      label string
    )
    STORED AS PARQUET
    LOCATION '{S3_DATA_PATH}'
    TBLPROPERTIES (
      'classification'='parquet',
      'compressionType'='snappy'
    );
"""

pd.read_sql(create_table_statement, conn)
```
